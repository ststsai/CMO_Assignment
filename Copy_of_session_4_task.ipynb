###`Problem 1:` Combine two lists index-wise(columns wise)

Write a program to add two lists index-wise. Create a new list that contains the 0th index item from both the list, then the 1st index item, and so on till the last element. any leftover items will get added at the end of the new list.

`Given List:`
```
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]
```

`Output:`
```
[['M','y'], ['na', me'], ['i', 's'], ['Kh', 'an']]
```

````python
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]

# Combine the lists index-wise using zip
combined_list = []
for item1, item2 in zip(list1, list2):
    combined_list.append([item1, item2])

# Handle any leftover items if lists were of different lengths, based on problem description.
# The provided example output implies a simple zip (truncating to the shorter list).
# If list1 and list2 were of different lengths, and leftover items should be appended individually,
# a different approach like itertools.zip_longest or manual iteration would be needed.
# For this problem, sticking to the output example which is covered by simple zip.

print(combined_list)
````

###`Problem 5:` You are given a list of integers. You are asked to make a list by running through elements of the list by adding all elements greater and itself.

i.e. Say given list is `[2,4,6,10,1]`
resultant list will be `[22,20,10,23]`.

For 1st element `2` ->> these are greater `(4+6+10)` values and `2` itself so on adding becomes `22`.

For 2nd element `4` ->> greater elements are `(6, 10)` and `4` itself, so on adding `20`

like wise for all other elememts.

`[2,4,6,10,1]`-->`[22,20,16,10,23]`




````python
S = []
list1 = [2, 4, 6, 10, 1]

for i in range(len(list1)):
    sum = list1[i]       # start with the element itself

    for j in range(len(list1)):
        if list1[i] < list1[j]:
            sum = sum + list1[j]

    S.append(sum)

print(S)
````

###`Problem 6:` Find list of common unique items from two list. and show in increasing order

`Input`

```
num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]
```

`Output:`
```
[34, 67, 89]
```


````python
num1 = [23, 45, 67, 78, 89, 34]
num2 = [34, 89, 55, 56, 39, 67]

common = []

for i in num1:
    for j in num2:
        if i == j:
            if i not in common:   # Avoid duplicates
                common.append(i)

common.sort()     # Sort in increasing order

print(common)
````

###`Problem 7:` Sort a list of alphanumeric strings based on product value of numeric character in it. If in any string there is no numeric character take it's product value as 1.

`Input:`
```
['1ac21', '23fg', '456', '098d','1','kls']
```

`Output:`
```
['456', '23fg', '1ac21', '1', 'kls', '098d']

```

````python
string = ['1ac21', '23fg', '456', '098d', '1', 'kls']

# Find product of digits for each string
product = []

for s in string:
    p = 1
    found = False

    for ch in s:
        if ch.isnumeric():
            p = p * int(ch)
            found = True

    if not found:
        p = 1

    product.append(p)

# Bubble Sort
n = len(string)

for i in range(n):
    for j in range(n - 1 - i):
        if product[j] > product[j + 1]:
            # Swap products
            product[j], product[j + 1] = product[j + 1], product[j]

            # Swap corresponding strings
            string[j], string[j + 1] = string[j + 1], string[j]

print(string)
````

### `Problem 12:` Write a program that can find the max number of each row of a matrix

**Example:**

Input:

```bash
[[1,2,3],[4,5,6],[7,8,9]]
```

Output:
```bash
[3,6,9]
```

````python
list = [[1,3,2,4],[4,5,6],[7,8,9]]
for i in range(len(list)):
    for j in range(len(list[i])):
        if j == 0:
            max = list[i][j]
        else:
            if list[i][j] > max:
                max = list[i][j]
    print(max)

````
